```python

# ============================================================
# DAETHON INTERNSHIP - TASK 1
# Exploratory Data Analysis on Superstore Retail Sales
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
import os
import warnings

from google.colab import files
from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

print("Libraries imported successfully!")

# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("\nUpload your Superstore CSV dataset:")
uploaded = files.upload()

file_name = next(iter(uploaded))

if file_name.lower().endswith(".csv"):
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))
elif file_name.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(io.BytesIO(uploaded[file_name]))
else:
    raise ValueError("Please upload a CSV or Excel dataset.")

df.columns = df.columns.astype(str).str.strip()

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

# ============================================================
# 3. DATASET OVERVIEW
# ============================================================

print("\n" + "=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

display(df.head())
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)

# ============================================================
# 4. DATA QUALITY CHECKS
# ============================================================

print("\n" + "=" * 60)
print("DATA QUALITY CHECKS")
print("=" * 60)

missing_before = df.isnull().sum()
print("Missing values by column:")
display(missing_before.to_frame("Missing Values"))

print("Exact duplicate rows:", df.duplicated().sum())

# ============================================================
# 5. DATA CLEANING
# ============================================================

print("\n" + "=" * 60)
print("DATA CLEANING")
print("=" * 60)

df = df.drop_duplicates().copy()

# Convert date columns to datetime
for col in df.columns:
    if "date" in col.lower():
        df[col] = pd.to_datetime(df[col], errors="coerce")

print("Exact duplicate rows removed.")
print("Dataset shape after cleaning:", df.shape)
print("\nRemaining missing values:")
display(df.isnull().sum().to_frame("Missing Values"))

# We intentionally do not fill missing values automatically.
# Their treatment depends on the dataset and analysis.

# ============================================================
# 6. AUTOMATIC COLUMN DETECTION
# ============================================================

def find_col(possible_names):
    for col in df.columns:
        if col.strip().lower() in possible_names:
            return col
    return None

sales_col = find_col({"sales", "total sales", "revenue"})
profit_col = find_col({"profit", "total profit"})
order_col = find_col({"order id", "order_id", "order number"})
customer_col = find_col({"customer id", "customer_id"})
category_col = find_col({"category", "product category"})
subcat_col = find_col({"sub-category", "sub category", "subcategory"})
region_col = find_col({"region"})
segment_col = find_col({"segment", "customer segment"})
product_col = find_col({"product name", "product"})
discount_col = find_col({"discount"})
quantity_col = find_col({"quantity", "qty"})
order_date_col = find_col({"order date", "order_date"})

print("\nDetected columns:")
for name, value in {
    "Sales": sales_col,
    "Profit": profit_col,
    "Order ID": order_col,
    "Customer ID": customer_col,
    "Category": category_col,
    "Sub-category": subcat_col,
    "Region": region_col,
    "Segment": segment_col,
    "Product": product_col,
    "Discount": discount_col,
    "Quantity": quantity_col,
    "Order Date": order_date_col
}.items():
    print(f"{name}: {value}")

# ============================================================
# 7. OVERALL BUSINESS PERFORMANCE
# ============================================================

print("\n" + "=" * 60)
print("OVERALL BUSINESS PERFORMANCE")
print("=" * 60)

if sales_col:
    print(f"Total Sales: {df[sales_col].sum():,.2f}")

if profit_col:
    print(f"Total Profit: {df[profit_col].sum():,.2f}")

if order_col:
    print("Unique Orders:", df[order_col].nunique())

if customer_col:
    print("Unique Customers:", df[customer_col].nunique())

if quantity_col:
    print("Total Quantity Sold:", df[quantity_col].sum())

# ============================================================
# 8. SALES AND PROFIT BY CATEGORY
# ============================================================

if category_col and sales_col:
    category_metrics = {sales_col: "sum"}
    if profit_col:
        category_metrics[profit_col] = "sum"

    category_analysis = (
        df.groupby(category_col)
          .agg(category_metrics)
          .sort_values(sales_col, ascending=False)
    )

    print("\nSALES BY CATEGORY")
    display(category_analysis)

    category_analysis[sales_col].plot(kind="bar")
    plt.title("Total Sales by Product Category")
    plt.xlabel("Category")
    plt.ylabel("Sales")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

    if profit_col:
        category_analysis[profit_col].plot(kind="bar")
        plt.title("Total Profit by Product Category")
        plt.xlabel("Category")
        plt.ylabel("Profit")
        plt.xticks(rotation=30)
        plt.tight_layout()
        plt.show()

# ============================================================
# 9. REGIONAL PERFORMANCE
# ============================================================

if region_col and sales_col:
    region_analysis = df.groupby(region_col)[sales_col].sum()
    region_analysis = region_analysis.sort_values(ascending=False)

    print("\nSALES BY REGION")
    display(region_analysis.to_frame("Total Sales"))

    region_analysis.plot(kind="bar")
    plt.title("Total Sales by Region")
    plt.xlabel("Region")
    plt.ylabel("Sales")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# ============================================================
# 10. MONTHLY SALES TREND
# ============================================================

if order_date_col and sales_col:
    monthly_df = df.dropna(subset=[order_date_col]).copy()

    monthly_df["Month"] = (
        monthly_df[order_date_col].dt.to_period("M")
    )

    monthly_sales = (
        monthly_df.groupby("Month")[sales_col]
        .sum()
        .sort_index()
    )

    print("\nMONTHLY SALES")
    display(monthly_sales.to_frame("Total Sales"))

    monthly_sales.index = monthly_sales.index.astype(str)

    monthly_sales.plot(kind="line", marker="o")
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Sales")
    plt.xticks(rotation=60)
    plt.tight_layout()
    plt.show()

# ============================================================
# 11. TOP 10 PRODUCTS
# ============================================================

if product_col and sales_col:
    top_products = (
        df.groupby(product_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print("\nTOP 10 PRODUCTS BY SALES")
    display(top_products.to_frame("Total Sales"))

    top_products.sort_values().plot(kind="barh")
    plt.title("Top 10 Products by Sales")
    plt.xlabel("Sales")
    plt.tight_layout()
    plt.show()

# ============================================================
# 12. CUSTOMER SEGMENT ANALYSIS
# ============================================================

if segment_col and sales_col:
    segment_sales = (
        df.groupby(segment_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nSALES BY CUSTOMER SEGMENT")
    display(segment_sales.to_frame("Total Sales"))

    segment_sales.plot(kind="bar")
    plt.title("Sales by Customer Segment")
    plt.xlabel("Segment")
    plt.ylabel("Sales")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

# ============================================================
# 13. DISCOUNT AND PROFIT ANALYSIS
# ============================================================

if discount_col and profit_col:
    discount_analysis = (
        df.groupby(discount_col)[profit_col]
        .mean()
        .sort_index()
    )

    print("\nAVERAGE PROFIT BY DISCOUNT LEVEL")
    display(discount_analysis.to_frame("Average Profit"))

    plt.figure()
    sns.scatterplot(
        data=df,
        x=discount_col,
        y=profit_col,
        alpha=0.5
    )
    plt.title("Discount vs Profit")
    plt.tight_layout()
    plt.show()

# ============================================================
# 14. SUB-CATEGORY PROFITABILITY
# ============================================================

if subcat_col and profit_col:
    subcat_profit = (
        df.groupby(subcat_col)[profit_col]
        .sum()
        .sort_values()
    )

    print("\nSUB-CATEGORY PROFITABILITY")
    display(subcat_profit.to_frame("Total Profit"))

    subcat_profit.plot(kind="barh")
    plt.title("Profit by Product Sub-category")
    plt.xlabel("Profit")
    plt.tight_layout()
    plt.show()

# ============================================================
# 15. CORRELATION ANALYSIS
# ============================================================

numeric_df = df.select_dtypes(include="number")

if numeric_df.shape[1] >= 2:
    correlation = numeric_df.corr()

    print("\nCORRELATION MATRIX")
    display(correlation)

    plt.figure(figsize=(10, 7))
    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )
    plt.title("Correlation Between Numerical Variables")
    plt.tight_layout()
    plt.show()

# ============================================================
# 16. KEY BUSINESS FINDINGS
# ============================================================

print("\n" + "=" * 60)
print("KEY BUSINESS FINDINGS")
print("=" * 60)

if category_col and sales_col:
    category_totals = df.groupby(category_col)[sales_col].sum()
    print("Highest-sales category:", category_totals.idxmax())
    print("Lowest-sales category:", category_totals.idxmin())

if region_col and sales_col:
    region_totals = df.groupby(region_col)[sales_col].sum()
    print("Highest-sales region:", region_totals.idxmax())

if segment_col and sales_col:
    segment_totals = df.groupby(segment_col)[sales_col].sum()
    print("Highest-sales segment:", segment_totals.idxmax())

if order_date_col and sales_col:
    valid_dates = df.dropna(subset=[order_date_col]).copy()

    if not valid_dates.empty:
        monthly_totals = valid_dates.groupby(
            valid_dates[order_date_col].dt.to_period("M")
        )[sales_col].sum()

        if not monthly_totals.empty:
            print("Best month:", str(monthly_totals.idxmax()))
            print(f"Best-month sales: {monthly_totals.max():,.2f}")

# ============================================================
# 17. EXPORT RESULTS
# ============================================================

# Save cleaned data
df.to_csv("superstore_cleaned.csv", index=False)

# Export key tables together into an Excel workbook
with pd.ExcelWriter("superstore_eda_results.xlsx") as writer:
    df.describe(include="all").T.to_excel(
        writer, sheet_name="Dataset Summary"
    )

    if category_col and sales_col:
        category_analysis.to_excel(
            writer, sheet_name="Category Analysis"
        )

    if region_col and sales_col:
        region_analysis.to_frame("Total Sales").to_excel(
            writer, sheet_name="Regional Analysis"
        )

    if order_date_col and sales_col:
        monthly_sales.to_frame("Total Sales").to_excel(
            writer, sheet_name="Monthly Sales"
        )

print("\nEDA COMPLETED SUCCESSFULLY!")
print("Created: superstore_cleaned.csv")
print("Created: superstore_eda_results.xlsx")

# Uncomment these lines to download the output files:
# files.download("superstore_cleaned.csv")
# files.download("superstore_eda_results.xlsx")
```
